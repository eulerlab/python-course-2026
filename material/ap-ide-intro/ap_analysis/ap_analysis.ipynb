# Debugging

The next three functions, `plot_mean_amplitude_v1`, `_v2`, `_v3`, are each a copy
of `plot_mean_amplitude` from Task 2 with **exactly one bug**. Your job is to find
the bug using the **debugger**, not by staring at the code.

**The three debugger tools you need today:** 
- **Breakpoint** — click in the margin just left of a line number; a red dot
  appears. When you run ther program in 'debug mode' the program will *pause* on that line so you can look around.
- **Variables panel** — when paused, it shows the current value of every
  variable. This is how you see what actually went wrong. Find the variables panel at the top left box when you are in 'debug mode'.
- **Debug console** - you can execute statements like `print(population_control.shape)` there and see what comes out. You can find this in the lower panel, next to the terminal.

To run a notebook cell with the debugger, use **"Debug Cell"** (the drop-down
arrow next to the Run button), not the normal Run.

You will see that a little bar at the top of the screen appears when you are in the debug mode. 
You can press the arrow on the left **Continue** to continue running the code until the next breakpoint. You can also just run the next line by pressing **Step Over** (run the current line).
